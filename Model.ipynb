## Feature Engineering

Implementing the feature engineering steps outlined in the instructions: handle missing 'Levy' values, calculating the car age and creating dummy variables for categorical features ensuring consistency between train and test sets, scale numerical features, and drop unnecessary columns.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

def preprocess_data(train_df, test_df):
    # Working on the copies to avoid modifying original data
    train_df = train_df.copy()
    test_df = test_df.copy()

    # 1. Handling the missing values in the 'Levy' column
    train_df['Levy'] = train_df['Levy'].replace('-', '0').astype(float)
    test_df['Levy'] = test_df['Levy'].replace('-', '0').astype(float)

    # 2. Extracting the age of the car
    current_year = 2025
    train_df['Age'] = current_year - train_df['Prod. year']
    test_df['Age'] = current_year - test_df['Prod. year']

    # 3. One-hot encode categorical variables
    categorical_cols = ['Manufacturer', 'Category', 'Fuel type', 'Gear box type', 
                        'Drive wheels', 'Wheel', 'Color', 'Doors']
    combined_df = pd.concat([train_df.drop('Price', axis=1), test_df], ignore_index=True)
    combined_df = pd.get_dummies(combined_df, columns=categorical_cols, dummy_na=False)

    # 4. Spliting back into train and test
    train_processed_df = combined_df.iloc[:len(train_df)].copy()
    test_processed_df = combined_df.iloc[len(train_df):].copy()
    train_processed_df['Price'] = train_df['Price'].values

    # 5. Scaling numerical features
    numerical_cols = ['Age', 'Engine volume_numeric', 'Mileage_numeric', 'Cylinders', 'Airbags', 'Levy']
    train_cols = [col for col in numerical_cols if col in train_processed_df.columns]
    test_cols = [col for col in numerical_cols if col in test_processed_df.columns]

    scaler = StandardScaler()
    if train_cols:
        train_processed_df[train_cols] = scaler.fit_transform(train_processed_df[train_cols])
    if test_cols:
        test_processed_df[test_cols] = scaler.transform(test_processed_df[test_cols])

    # 6. Droping original and unnecessary columns
    drop_cols = ['Engine volume', 'Mileage', 'Prod. year', 'ID']
    train_processed_df = train_processed_df.drop(columns=[col for col in drop_cols if col in train_processed_df.columns])
    test_processed_df = test_processed_df.drop(columns=[col for col in drop_cols if col in test_processed_df.columns])

    return train_processed_df, test_processed_df

# Loading the original data
original_train_df = pd.read_csv('train.csv')  # Change path if needed
original_test_df = pd.read_csv('test.csv')    # Change path if needed

# Preprocessing
train_df, test_df = preprocess_data(original_train_df, original_test_df)

# Displaying results
print("Processed Train DataFrame Head:")
display(train_df.head())
print("\nProcessed Train DataFrame Shape:", train_df.shape)

print("\nProcessed Test DataFrame Head:")
display(test_df.head())
print("\nProcessed Test DataFrame Shape:", test_df.shape)
```

## Model Optimization

Identifing the best performing models based on the evaluation metrics on the training data and select Gradient Boosting and Random Forest for hyperparameter tuning due to their good performance and potential for less overfitting compared to the Decision Tree and defing parameter grids for these models and using GridSearchCV to perform hyperparameter tuning with cross-validation.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
import warnings

warnings.filterwarnings('ignore', category=FutureWarning)

# Gradient Boosting and Random Forest show strong performance and are good candidates for tuning.

# 1. Choose models for hyperparameter tuning
gbr_tuned = GradientBoostingRegressor(random_state=42)
rf_tuned = RandomForestRegressor(random_state=42)

# 2. Define parameter grids for the chosen models

# Parameter grid for Gradient Boosting
param_grid_gbr = {
    'n_estimators': [100, 200, 300],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 4, 5],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Parameter grid for Random Forest
param_grid_rf = {
    'n_estimators': [100, 200, 300],
    'max_depth': [10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# 3. Using GridSearchCV to perform hyperparameter tuning with cross-validation

# GridSearchCV for Gradient Boosting
grid_search_gbr = GridSearchCV(estimator=gbr_tuned, param_grid=param_grid_gbr,
                               scoring='neg_mean_squared_error', cv=3, n_jobs=-1, verbose=2)

# GridSearchCV for Random Forest
grid_search_rf = GridSearchCV(estimator=rf_tuned, param_grid=param_grid_rf,
                              scoring='neg_mean_squared_error', cv=3, n_jobs=-1, verbose=2)

# 4. Fit the tuning process to the training data
print("Starting GridSearchCV for Gradient Boosting...")
grid_search_gbr.fit(X_train, y_train)
print("Finished GridSearchCV for Gradient Boosting.")

print("\nStarting GridSearchCV for Random Forest...")
grid_search_rf.fit(X_train, y_train)
print("Finished GridSearchCV for Random Forest.")

# 5. Extracting the best parameters found
best_params_gbr = grid_search_gbr.best_params_
best_params_rf = grid_search_rf.best_params_

print("\nBest parameters for Gradient Boosting:", best_params_gbr)
print("Best parameters for Random Forest:", best_params_rf)

# 6. Training new instances of the chosen models using the best parameters found
optimized_gbr = GradientBoostingRegressor(**best_params_gbr, random_state=42)
optimized_rf = RandomForestRegressor(**best_params_rf, random_state=42)

print("\nTraining optimized Gradient Boosting model...")
optimized_gbr.fit(X_train, y_train)
print("Optimized Gradient Boosting model trained.")

print("Training optimized Random Forest model...")
optimized_rf.fit(X_train, y_train)
print("Optimized Random Forest model trained.")

# 7. Evaluating the optimized models on the training data
y_train_pred_optimized_gbr = optimized_gbr.predict(X_train)
y_train_pred_optimized_rf = optimized_rf.predict(X_train)

metrics_optimized = {}

metrics_optimized['Optimized Gradient Boosting'] = {
    'MAE': mean_absolute_error(y_train, y_train_pred_optimized_gbr),
    'MSE': mean_squared_error(y_train, y_train_pred_optimized_gbr),
    'R-squared': r2_score(y_train, y_train_pred_optimized_gbr)
}

metrics_optimized['Optimized Random Forest'] = {
    'MAE': mean_absolute_error(y_train, y_train_pred_optimized_rf),
    'MSE': mean_squared_error(y_train, y_train_pred_optimized_rf),
    'R-squared': r2_score(y_train, y_train_pred_optimized_rf)
}

# Storing optimized metrics in a DataFrame and display
metrics_optimized_df = pd.DataFrame(metrics_optimized).T
print("\nEvaluation Metrics on Training Data (Optimized Models):")
display(metrics_optimized_df)

# Comparing with original metrics
print("\nEvaluation Metrics on Training Data (Original Models):")
display(metrics_df)
```
